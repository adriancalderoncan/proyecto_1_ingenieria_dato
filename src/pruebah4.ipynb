```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Carpeta principal del proyecto, independientemente de dónde ejecutes Python
PROYECTO = Path(r"../data/HU3_medicamentos_diabetes.xlsx").resolve().parents[1]
ENTRADA = PROYECTO / "data" / "HU3_medicamentos_diabetes.xlsx"
SALIDA = PROYECTO / "data" / "HU4_medicamentos_diabetes_limpio.csv"
FIGURAS = PROYECTO / "reports" / "figures"

FIGURAS.mkdir(parents=True, exist_ok=True)
```

```python
df = pd.read_excel(ENTRADA)

print("Filas y columnas:", df.shape)
print("\nColumnas:")
print(df.columns.tolist())
print("\nTipos:")
print(df.dtypes)
print("\nValores ausentes:")
print(df.isna().sum())
print("\nRegistros duplicados:", df["nregistro"].duplicated().sum())
```

```python
assert len(df.columns) == 32, "Revisa el Excel: no tiene las 32 columnas esperadas"

df.columns = list(df.columns[:26]) + [
    "codigo_nacional_nomenclator",
    "estado_nomenclator",
    "precio_pvp_iva",
    "precio_referencia",
    "tratamiento_larga_duracion",
    "especial_control_medico",
]
```

```python
df.describe()
```

```python
# Son identificadores, aunque en Excel parezcan números.
df["nregistro"] = df["nregistro"].astype("string")
df["cn"] = df["cn"].astype("Int64").astype("string")
df["codigo_nacional_nomenclator"] = (
    df["codigo_nacional_nomenclator"].astype("Int64").astype("string")
)

# CIMA guardó estas fechas como milisegundos desde 1970.
for columna in ["estado_aut", "estado_rev"]:
    df[columna] = pd.to_datetime(df[columna], unit="ms", errors="coerce")

# En tres filas aparece el texto "no tiene sección 4.4".
# Lo convertimos en un valor ausente, no en cero palabras.
df["palabras_seccion_4_4"] = pd.to_numeric(
    df["palabras_seccion_4_4"], errors="coerce"
).astype("Int64")

for columna in ["precio_pvp_iva", "precio_referencia"]:
    df[columna] = pd.to_numeric(df[columna], errors="coerce")
```

```python
con_cruce = df["codigo_nacional_nomenclator"].notna()

assert df.loc[con_cruce, "cn"].eq(
    df.loc[con_cruce, "codigo_nacional_nomenclator"]
).all(), "Hay códigos nacionales que no coinciden"

df = df.drop(columns="codigo_nacional_nomenclator")
```

```python
print("\nResumen de variables numéricas:")
print(
    df[
        [
            "precio_pvp_iva",
            "precio_referencia",
            "palabras_seccion_4_4",
            "numero_tablas",
            "indicador_riesgo_severo",
            "num_excipientes",
        ]
    ].describe()
)

print("\nMedicamentos por forma farmacéutica:")
print(df["forma_farmaceutica_simplificada"].value_counts().head(10))

print("\nPrecio según si es genérico:")
print(
    df.groupby("generico")["precio_pvp_iva"]
    .agg(["count", "median", "mean"])
)
```

```python
sns.histplot(df["precio_pvp_iva"].dropna(), bins=30)
plt.title("Distribución del precio PVP con IVA")
plt.xlabel("Precio (€)")
plt.tight_layout()
plt.savefig(FIGURAS / "precio_pvp.png")
plt.close()

sns.histplot(df["palabras_seccion_4_4"].dropna(), bins=30)
plt.title("Palabras en la sección 4.4")
plt.xlabel("Número de palabras")
plt.tight_layout()
plt.savefig(FIGURAS / "palabras_seccion_4_4.png")
plt.close()

df["forma_farmaceutica_simplificada"].value_counts().head(10).sort_values().plot(
    kind="barh"
)
plt.title("Diez formas farmacéuticas más frecuentes")
plt.xlabel("Número de medicamentos")
plt.tight_layout()
plt.savefig(FIGURAS / "formas_farmaceuticas.png")
plt.close()

sns.boxplot(
    data=df[df["precio_pvp_iva"].notna()],
    x="generico",
    y="precio_pvp_iva",
    showfliers=False,
)
plt.title("Precio PVP de genéricos y no genéricos")
plt.xlabel("Genérico: 0 = no, 1 = sí")
plt.ylabel("Precio (€)")
plt.tight_layout()
plt.savefig(FIGURAS / "precio_genericos.png")
plt.close()
```

```python
for columna in [
    "precio_pvp_iva",
    "precio_referencia",
    "palabras_seccion_4_4",
    "numero_tablas",
    "indicador_riesgo_severo",
]:
    negativos = (df[columna].dropna() < 0).sum()
    print(f"{columna}: {negativos} valores negativos")

for columna in [
    "comercializado",
    "requiere_receta",
    "generico",
    "afecta_conduccion",
    "triangulo_negro",
    "medicamento_huerfano",
    "biosimilar",
]:
    incorrectos = (~df[columna].isin([0, 1])).sum()
    print(f"{columna}: {incorrectos} valores fuera de 0 y 1")
```

```python
assert len(df) == 1569
assert df["nregistro"].is_unique

df.to_csv(SALIDA, index=False, encoding="utf-8-sig")

comprobacion = pd.read_csv(SALIDA)
assert len(comprobacion) == len(df)
assert comprobacion["nregistro"].is_unique

print(f"\nCSV guardado en: {SALIDA}")
print("Dimensiones finales:", comprobacion.shape)
```
