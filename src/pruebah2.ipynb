```python
import requests
from bs4 import BeautifulSoup

df = pd.read_excel("../data/HU1_medicamentos_diabetes.xlsx")

url = df["url_html_ficha_tecnica"].iloc[0]      #la url de la primera fila de la columna url_html_ficha_tecnica
response = requests.get(url, timeout=30)

print(response.status_code)   # atributo: 200 = OK
response.raise_for_status()   # método: si hubiese fallado (4xx/5xx), para aquí con un error claro
```

```python
html_parseado = BeautifulSoup(response.text, "html.parser")

apartado_4_4 = html_parseado.find(id="4.4")
print(apartado_4_4)

#dentro de la estructura del html(ver el link,y darle a inspeccionar,vemos que dentro h2 solo contiene el id. )
#el contenido(texto) esta dentro de la siguiente etiqueta,que es un <div>.Esta etiqueta es "hermana"(esta al mismo nivel)
#por tanto,para acceder a ella,usamos el metodo find_next_sibling de BeautifulSoup

contenido_4_4 = apartado_4_4.find_next_sibling() #nos da el contenido del <div>
print(contenido_4_4)
#vemos que el contenido esta contenido dentro de diferentes parrafos (<p>).Ahora tenemos que obtener ese texto para contar las palabras
```

```python
texto_4_4 = contenido_4_4.get_text(" ",strip=True)
#get_text() coge todo el texto visible que hay dentro de ese elemento y de todo lo que cuelga de él (los <p>, los <span>...), ignorando las etiquetas.
#strip=True quita espacios sobrantes al principio y al final.
#El primer argumento, " " (un espacio), le dice cómo unir el texto cuando viene de varias etiquetas distintas — así evitamos que dos palabras de párrafos distintos se te peguen sin espacio entre medias.
print(texto_4_4)

numero_palabras = len(texto_4_4.split())
print(f"el numero de palabras es {numero_palabras}")
#.split() sin nada dentro parte el texto por cualquier espacio en blanco (espacios, saltos de línea...) y de paso ignora los espacios de más
```
