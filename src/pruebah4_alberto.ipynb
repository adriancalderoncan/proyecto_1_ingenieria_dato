```python
# Tras estudiar en la celda anterior todos los posibles datos distintos que aparecian en las dosis de
# nuestros medicamentos, decidimos utilizar la IA generativa para poder obtener una fila resultante para
# cada caso concreto de nuestro dataset. El objetivo es poder obtener las dosis de los medicamentos 
# estandarizadas

# ============================================================
# LIMPIEZA Y ESTANDARIZACIÓN DE LA COLUMNA "dosis"
# Objetivo: dejar todas las dosis con el formato "número espacio unidad"
# (o "número unidad/número unidad" en combinaciones de dos principios activos)
# y después extraer el número.
# IMPORTANTE: esta celda va después de convertir nregistro a texto (paso 8).
# ============================================================


# PASO 1: corregir a mano las filas con valores desconocidos, incompletos o erróneos.
# El mismo texto ("Desconocido", "50 - REVISAR mg") corresponde a dosis distintas
# según la fila, así que no se puede reemplazar por texto: corregimos fila a fila usando nregistro.
# Justificación: la dosis correcta se ha sacado del nombre oficial del medicamento.
# ------------------------------------------------------------
correcciones_por_fila = {
    "150001": "15 mg",               # ACTOS 15 mg (ponía "Desconocido")
    "150009": "15 mg",               # ACTOS 15 mg (ponía "Desconocido")
    "150004": "30 mg",               # ACTOS 30 mg (ponía "Desconocido")
    "150010": "30 mg",               # ACTOS 30 mg (ponía "Desconocido")
    "98070001": "75 mg",             # ISCOVER 75 MG (ponía "Desconocido")
    "134006": "100 UI/ml",           # LANTUS 100 UNIDADES/ML (ponía "Desconocida")
    "1181265011": "2,5 mg/1000 mg",  # SEGLUROMET 2,5 MG/1.000 MG (ponía "2.5 - REVISAR mg")
    "1181265025": "7,5 mg/1000 mg",  # SEGLUROMET 7,5 MG/1.000 MG (ponía "7.5 - REVISAR mg")
    "86341": "50 mg/1000 mg",        # SITAGLIPTINA/METFORMINA STADA 50 MG/1.000 MG (ponía "50 - REVISAR mg")
    "86283": "50 mg/1000 mg",        # VILDAGLIPTINA/METFORMINA STADA 50 MG/1000 MG (ponía "50 - REVISAR mg")
    "86284": "50 mg/850 mg",         # VILDAGLIPTINA/METFORMINA STADA 50 MG/850 MG (ponía "50 - REVISAR mg")
    "9529002": "6 mg/ml",            # VICTOZA 6 MG/ML (ponía "60 mg", que es un error: la ficha técnica dice 6 mg por ml)
}

# Recorremos el diccionario y, para cada nregistro, sustituimos su dosis por la correcta
for nregistro, dosis_correcta in correcciones_por_fila.items():
    df_clean.loc[df_clean["nregistro"] == nregistro, "dosis"] = dosis_correcta


# ------------------------------------------------------------
# PASO 2: pasar a minúsculas y quitar espacios sobrantes.
# Así "50 MG" y "50 mg" quedan iguales y desaparecen los espacios dobles y los saltos de línea (\n).
# ------------------------------------------------------------
df_clean["dosis"] = df_clean["dosis"].str.lower()

# split() parte el texto por cualquier espacio, salto de línea o espacio doble,
# y join(" ") lo vuelve a unir con un solo espacio entre palabras
df_clean["dosis"] = df_clean["dosis"].str.split().str.join(" ")


# ------------------------------------------------------------
# PASO 3: quitar los nombres de principios activos y el texto sobrante.
# El principio activo ya está en la columna "pactivos", así que en "dosis" sobra.
# Ponemos primero los textos más largos: si quitáramos antes "losartan",
# en "losartan potasico" quedaría suelto "potasico".
# ------------------------------------------------------------
quitar = [
    "como hidrogenosulfato",
    "metformina hidrocloruro",
    "glucagón hidrocloruro",
    "losartan potasico",
    "losartan potasio",
    "losartan",
    "lisinopril dihidrato",
    "lisinopril",
    "simvastatina",
    "irbesartan",
    "telmisartan",
    "rivaroxaban",
    "pioglitazona",
    "glimepirida",
    "captopril",
    "clopidogrel",
    "clopidrogel",                         # errata del original
    "clortalidona",
    "ramipril,",                           # "Ramipril, 2,5 mg" (también hay que quitar la coma)
    "ramipril",
    "acarbosa",
    "sitagliptina",
    "linagliptina",
    "linagiptina",                         # errata del original
    "saxagliptina",
    "(empagliflozin/linagliptin) mg/mg",
    "/ comprimido",
    "/comprimido",
    "comprimidos",
    " de ",                                # "30 mg de pioglitazona" -> "30 mg"
]

# Recorremos la lista y vamos borrando cada texto (lo cambiamos por un espacio)
for texto in quitar:
    df_clean["dosis"] = df_clean["dosis"].str.replace(texto, " ", regex=False)

# Al borrar quedan espacios dobles o al principio/final: los limpiamos otra vez
df_clean["dosis"] = df_clean["dosis"].str.split().str.join(" ")


# ------------------------------------------------------------
# PASO 4: corregir casos concretos con un diccionario.
# Son valores únicos que no siguen ninguna regla general; cada línea es un caso justificado.
# ------------------------------------------------------------
casos_concretos = {
    "75 mh": "75 mg",                    # errata: mh -> mg
    "30 gm + 2 mg": "30 mg/2 mg",        # errata: gm -> mg (TANDEMACT 30 mg/2 mg)
    "1000/50 mg/mg": "50 mg/1000 mg",    # orden invertido (RISTFOR, VELMETIA, ZOMARIST 50 MG/1000 MG)
    "850/50 mg/mg": "50 mg/850 mg",      # orden invertido (ZOMARIST 50 mg/850 mg)
    "50/1000 mg": "50 mg/1000 mg",       # falta la unidad en el primer número
    "50/850 mg": "50 mg/850 mg",         # falta la unidad en el primer número
    "15": "15 mg",                       # sin unidad (RIXACAM 15 MG)
    "20": "20 mg",                       # sin unidad (RIXACAM 20 MG)
    "0,5 mg desmopresina acetato/ 0,5 ml": "0,1 mg/ml",  # el nombre del medicamento dice 0,1 mg/ml
    "10 microgramos": "0,01 mg",         # 1 mg = 1000 microgramos
    "20 microgramos": "0,02 mg",
    "120 microgramos": "0,12 mg",
    "4 µg/ml": "0,004 mg/ml",            # µg es otra forma de escribir microgramos
}

# replace con un diccionario cambia los valores que coinciden EXACTAMENTE con una clave
df_clean["dosis"] = df_clean["dosis"].replace(casos_concretos)


# ------------------------------------------------------------
# PASO 5: poner un espacio entre el número y "mg" ("15mg" -> "15 mg").
# Es una expresión regular: (\d) significa "un número" y \1 es "ese mismo número".
# Se lee: "donde haya un número pegado a mg, escribe el número, un espacio y mg".
# ------------------------------------------------------------
df_clean["dosis"] = df_clean["dosis"].str.replace(r"(\d)mg", r"\1 mg", regex=True)


# ------------------------------------------------------------
# PASO 6: usar "/" como único separador de combinaciones.
# Las combinaciones de dos principios activos venían separadas con "+", "y", "-"
# o con espacios alrededor de "/".
# ------------------------------------------------------------
separadores = {
    " + ": "/",     # "30 mg + 4 mg"    -> "30 mg/4 mg"
    " y ": "/",     # "50 mg y 1000 mg" -> "50 mg/1000 mg"
    "-": "/",       # "15 mg-850 mg"    -> "15 mg/850 mg"
    " / ": "/",     # "15 mg / 850 mg"  -> "15 mg/850 mg"
    "/ ": "/",      # "50 mg/ 850 mg"   -> "50 mg/850 mg"
    " /": "/",      # "5 mg /850 mg"    -> "5 mg/850 mg"
}
for viejo, nuevo in separadores.items():
    df_clean["dosis"] = df_clean["dosis"].str.replace(viejo, nuevo, regex=False)


# ------------------------------------------------------------
# PASO 7: unificar el separador decimal.
# Python solo entiende el punto como decimal. Hay que quitar PRIMERO el punto de miles
# de "1.000"; si no, después se leería como 1.0.
# ------------------------------------------------------------
# 1) "1.000" es mil (punto de miles), no uno con decimales
df_clean["dosis"] = df_clean["dosis"].str.replace("1.000", "1000", regex=False)

# 2) La coma decimal española pasa a punto: "2,5 mg" -> "2.5 mg"
df_clean["dosis"] = df_clean["dosis"].str.replace(",", ".", regex=False)

# 3) "0.50 mg" es lo mismo que "0.5 mg"
df_clean["dosis"] = df_clean["dosis"].str.replace("0.50 mg", "0.5 mg", regex=False)


# ------------------------------------------------------------
# PASO 8: unificar las unidades de insulina.
# "U/ml", "UI/ml" y "unidades/ml" significan lo mismo: unidades internacionales por ml.
# ------------------------------------------------------------
df_clean["dosis"] = df_clean["dosis"].str.replace("unidades/ml", "UI/ml", regex=False)
df_clean["dosis"] = df_clean["dosis"].str.replace("ui/ml", "UI/ml", regex=False)
df_clean["dosis"] = df_clean["dosis"].str.replace("u/ml", "UI/ml", regex=False)


# ------------------------------------------------------------
# PASO 9: marcar qué dosis están expresadas "por ml" (líquidos e insulinas).
# Lo hacemos ANTES de pasarlo todo a mg para no perder esta información:
# en un comprimido los mg son por comprimido; en un líquido, por cada ml.
# ------------------------------------------------------------
df_clean["dosis_por_ml"] = 0
df_clean.loc[df_clean["dosis"].str.contains("/ml"), "dosis_por_ml"] = 1


# ------------------------------------------------------------
# PASO 10: pasar las insulinas de UI/ml a mg/ml.
# Las UI (unidades internacionales) miden actividad, no peso, y la equivalencia depende
# del tipo de insulina. La sacamos de la sección 2 de la ficha técnica de cada una
# ("1 ml contiene 100 unidades de insulina ... (equivalente a X mg)").
# Las fichas de Humulina no indican los mg; usamos la equivalencia de la insulina humana
# de Actrapid, Insulatard y Mixtard (100 UI = 3,5 mg).
# ------------------------------------------------------------
# Diccionario: principio activo -> mg que hay en 100 UI
mg_por_100_ui = {
    "INSULINA ASPARTA": 3.5,                                                   # Fiasp, NovoRapid
    "INSULINA ASPARTA, INSULINA ASPARTA PROTAMINA+": 3.5,                      # NovoMix
    "INSULINA DEGLUDEC": 3.66,                                                 # Tresiba
    "INSULINA DETEMIR": 14.2,                                                  # Levemir
    "INSULINA GLARGINA": 3.64,                                                 # Lantus, Abasaglar, Semglee, Toujeo
    "INSULINA GLULISINA": 3.49,                                                # Apidra
    "INSULINA LISPRO": 3.5,                                                    # Humalog
    "INSULINA LISPRO, INSULINA LISPRO PROTAMINA": 3.5,                         # Humalog Mix
    "INSULINA HUMANA": 3.5,                                                    # Humulina Regular, Humulina 30:70
    "INSULINA HUMANA ISOFANA": 3.5,                                            # Insulatard
    "INSULINA ISOFANA HUMANA PRB": 3.5,                                        # Humulina NPH
    "INSULINA SOLUBLE NEUTRA": 3.5,                                            # Actrapid
    "INSULINA SOLUBLE NEUTRA HUMANA PRB, INSULINA ISOFANA HUMANA PRB": 3.5,    # Mixtard
}

# Filas que son insulina (su dosis está en UI/ml)
es_insulina = df_clean["dosis"].str.contains("UI/ml")

# Número de UI por ml: el texto antes del espacio. "100 UI/ml" -> 100
ui_por_ml = df_clean.loc[es_insulina, "dosis"].str.split(" ").str[0].astype(float)

# Buscamos en el diccionario los mg por 100 UI de cada insulina según su principio activo
factor = df_clean.loc[es_insulina, "pactivos"].map(mg_por_100_ui)
print("Insulinas sin equivalencia en el diccionario:", factor.isna().sum())   # tiene que salir 0

# Regla de tres: si 100 UI son "factor" mg, entonces "ui_por_ml" UI son ui_por_ml * factor / 100 mg
mg_por_ml = (ui_por_ml * factor / 100).round(2)

# Escribimos el resultado con el formato "número mg/ml".
# "{:g}".format quita los decimales que sobran (7.0 -> "7", 3.66 -> "3.66")
df_clean.loc[es_insulina, "dosis"] = mg_por_ml.map("{:g}".format) + " mg/ml"


# ------------------------------------------------------------
# PASO 11: dejar todo como "mg".
# Ya está todo en mg o en mg/ml. Quitamos el "/ml" para que todas las dosis tengan el mismo formato;
# la columna dosis_por_ml (paso 9) guarda que esas dosis son por cada ml.
# ------------------------------------------------------------
df_clean["dosis"] = df_clean["dosis"].str.replace("mg/ml", "mg", regex=False)


# ------------------------------------------------------------
# COMPROBACIÓN: todas las dosis deben ser "número mg" o "número mg/número mg".
# ------------------------------------------------------------
pd.set_option("display.max_rows", None)          # mostrar todas las filas, sin "..."
print(df_clean["dosis"].value_counts().sort_index())
pd.reset_option("display.max_rows")               # volvemos al límite por defecto

# Contamos las filas que NO terminan en "mg" (tiene que salir 0)
print("Filas que no están en mg:", (~df_clean["dosis"].str.endswith(" mg")).sum())


# ------------------------------------------------------------
# PASO 12: quedarnos con el número de la dosis.
# En las combinaciones guardamos también el segundo número para no perder información.
# ------------------------------------------------------------
# Número de la dosis: el texto antes del primer espacio. "50 mg/1000 mg" -> "50"
df_clean["dosis_numero"] = df_clean["dosis"].str.split(" ").str[0].astype(float)

# Segundo número (solo en combinaciones como "50 mg/1000 mg" -> 1000).
# Cogemos lo que hay después de "/" y luego lo que hay antes del espacio.
# Si no hay "/", str[1] da nulo: to_numeric lo deja como NaN y fillna lo pasa a 0.
segunda_parte = df_clean["dosis"].str.split("/").str[1].str.split(" ").str[0]
df_clean["dosis_numero_2"] = pd.to_numeric(segunda_parte, errors="coerce").fillna(0)

# Comprobamos el resultado
df_clean[["nombre", "dosis", "dosis_numero", "dosis_numero_2", "dosis_por_ml"]].head(10)
```
